```python
#Gauss Legendre Quadrature Method

n_1=[-0.8611363116, -0.3399810436, 0.3399810436, 0.8611363116]
w_1=[0.34785484510, 0.6521451549, 0.6521451549, 0.3478548451]

def f(x):
    return x**2-3

def gauss_quadrature(a,b):
    s=0
    for i in range(len(n_1)):
        g=(b-a)/2
        h=(a+b)/2
        u=f((g*n_1[i])+h)
        v=w_1[i]*u
        s+=v

    I=s*g
    print(f'{I} is the result of integration limit from {a} to {b}')

gauss_quadrature(2,5)
```

```python
#Gauss-Lagaurre Method
import math as m
x_3=[0.415775,2.29428,6.28995]
w_3=[0.711093,0.278518,0.010389]

x_4=[0.322548,1.74576,4.53662,9.39507]
w_4=[0.603154,0.357419,0.038888,0.000539]

def f(x):
    return m.exp(-x)*(x**3+2*x+1)

# f(x), limit from 0 to infinity

I=0
for i in range(len(x_3)):
    s=w_3[i]*m.exp(x_3[i])*f(x_3[i])
    I+=s
print(f'The integration is {I}')

# f(x), limit from a to infinity

I=0
a=int(input('Enter the lower limit(a): '))
for i in range(len(x_4)):
    s=w_4[i]*m.exp(x_4[i])*f(x_4[i]+a)
    I+=s
print(f'The integration is {I}')
```

```python
#Gauss Hermite Quadrature Method

x_3=[-1.224745,0,1.224745]
w_3=[0.295409,1.181636,0.295409]

def f(x):
    return x**2+2

I=0
for i in range(len(x_3)):
    s=w_3[i]*f(x_3[i])
    I+=s
print(f'The integration of the given function is {I}')
```
