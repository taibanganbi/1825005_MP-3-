Aim: To determine nth roots of a complex number and represent it in polar form

Formula Used: $$z=r^\frac{1}{n}\left[cos\left(\frac{2\pi k+\theta}{n}\right)+isin\left(\frac{2\pi k+\theta}{n}\right)\right]$$

```python
#nth roots of a complex number and representation in Polar form

import cmath as cm
import math as m

n=int(input('Enter the value of n: '))
z=1+1j
x=z.real
y=z.imag
r=m.sqrt(x**2+y**2)
theta=m.atan(y/x)
p=m.pi

print('The nth roots in polar form are: ')

for i in range(n):
    angle=(2*i*p+theta)/n
    a=round(angle,3)

    print(f'\nz_{i+1} ={round(r**(1/n),3)} cos({a})+isin({a})')
```

Aim: Translation of complex numbers as 2D vectors

Formula Used: $$w=f(z)=z+c$$ where, $c=a+ib$


```python
#Translation


import cmath as cm
import math as m

c=4+2j
x=[0,5,10,15,20]
y=[0,5,10,15,20]
x_list=[]
y_list=[]
u_list=[]
v_list=[]

for i in range(len(x)):
    z=complex(2*x[i],3*y[i])
    w=z+c
    u=w.real
    v=w.imag
    u_list.append(u)
    v_list.append(v)
    x_list.append(2*x[i])
    y_list.append(3*y[i])

import matplotlib.pyplot as plt
plt.plot(x_list,y_list)
plt.plot(u_list,v_list)
plt.xlabel('u')
plt.ylabel('v')
plt.legend(['z','w'])
plt.grid()
plt.show()
```

Aim: Rotation of complex numbers as 2D vectors

Formula Used:
$$w=f(z)=ze^{i\theta}$$

```python
#Rotation
import math as m
import cmath as cm

x=[0,1,1]
y=[1,1,-1]
x_list=[]
y_list=[]
u_list=[]
v_list=[]

for i in range(len(x)):
    z=complex(4*x[i],3*y[i])
    theta=5*m.pi/3
    w=z*(complex(m.cos(theta),m.sin(theta)))
    u=w.real
    v=w.imag
    u_list.append(u)
    v_list.append(v)
    x_list.append(4*x[i])
    y_list.append(3*y[i])

import matplotlib.pyplot as plt
plt.plot(x,y)
plt.plot(u_list,v_list)
plt.xlabel('u')
plt.ylabel('v')
plt.grid()
plt.show()
```

Aim: Reflection of complex numbers as 2D vectors

Formula Used: $$w=f(z)=-z$$


```python
#Reflection

import math as m
import cmath as cm

x=[0,5,10,15,20]
y=[0,5,10,15,20]
x_list=[]
y_list=[]
u_list=[]
v_list=[]

for i in range(len(x)):
    z=complex(2*x[i],y[i])
    w=z.conjugate()
    u=w.real
    v=w.imag
    u_list.append(u)
    v_list.append(v)
    x_list.append(2*x[i])
    y_list.append(y[i])

import matplotlib.pyplot as plt
plt.plot(x_list,y_list)
plt.plot(u_list,v_list)
plt.xlabel('Real Axis')
plt.ylabel('Imaginary Axis')
plt.grid()
plt.legend(['z','f(z)=-z'])
plt.show()
```

Aim: To visualise of mappings of elementary complex functions w=f(z) from z-plane to w-plane

Formula used:
$$w=f(z)=u+iv$$

```python
#Mappings of elementary complex functions w=f(z) from z-plane to w-plane
#f(z)=z^2

x=[-2,-1,0,1,2]
y=[0,1,2,3,4]
x_list=[]
y_list=[]
u_list=[]
v_list=[]

for i in range(len(x)):
    z=complex(2*x[i],y[i])
    f_z=z**2
    u=f_z.real
    u_list.append(u)
    v=f_z.imag
    v_list.append(v)
    x_list.append(2*x[i])
    y_list.append(y[i])

import matplotlib.pyplot as plt
plt.plot(x_list,y_list)
plt.plot(u_list,v_list)
plt.xlabel('Real Axis')
plt.ylabel('Imaginary Axis')
plt.legend(['z','f(z)'])
plt.grid()
plt.show()
```
