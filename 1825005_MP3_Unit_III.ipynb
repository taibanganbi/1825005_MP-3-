Aim: To compute DFT using complex numbers

Formula Used: $$F_k=\sum_{n=0}^{N-1} f(n)e^\frac{-i2\pi kn}{N}$$ where, $k=$no. of rows $$$$ $n=$ no. of columns


```python
#Discrete Fourier Transform(using complex number, 4 point DFT)
import numpy as np
import cmath as cm

f_k=np.array([1,2,3,4])
N=len(f_k)
w_m=np.zeros((N,N),dtype=complex)
p=cm.pi

for k in range(N):
    for n in range(N):
        c=round(np.cos(2*p*k*n/N),4)
        s=round(-np.sin(2*p*k*n/N),4)
        a=complex(c,s)
        w_m[k,n]=a

print(f'DFT Matrix:\n{w_m}')
b=np.matmul(w_m,f_k)
print('\nThe DFT coefficients are: ')

for i in range(len(b)):
   print(f'F_{i}={b[i]}')
```
