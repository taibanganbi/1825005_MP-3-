Aim: To find solution of ordinary differential equation(BVP) by Finite Difference Method

Formula Used: $$\begin{pmatrix}
-2+\beta_1 d^2 & 1+\frac{\alpha_1 d}{2} & 0 & 0 &0 ....\\
1-\frac{\alpha_2 d}{2} & -2+\beta_2 d^2 & 1+\frac{\alpha_2 d}{2}& 0 &0 ...\\
0& 1-\frac{\alpha_3 d}{2} & -2+\beta_3 d^2 &1+\frac{\alpha_3d}{2} & 0..\\
. &.&.&.&.\\
.&.&.&.&.\\
0&.&.&1-\frac{\alpha_N d}{2}& -2+\beta_N d^2\end{pmatrix}$$

```python
#Finite Difference Method
import numpy as np

x_1=0
x_n=4
y_0=1
y_n=2
n=int(input('Enter the value of n: '))
d=(x_n-x_1)/n
alpha=2
beta=0
gamma=1
m=n-1
matrix_1=np.zeros((m,m))

for i in range(m):
    for j in range(m):
        if i==j:
            a=-2+beta*d**2
            matrix_1[i,j]=a
        elif j==i+1:
            b=1+alpha*d/2
            matrix_1[i,j]=b
        elif j==i-1:
            c=1-alpha*d/2
            matrix_1[i,j]=c

matrix_2=np.linalg.inv(matrix_1)
t=gamma*d**2
matrix_3=np.full(m,t)
matrix_3[0] -= c*y_0
matrix_3[-1] -= b*y_n
y=np.matmul(matrix_2,matrix_3)

print(f'{y} is the solution')
```
