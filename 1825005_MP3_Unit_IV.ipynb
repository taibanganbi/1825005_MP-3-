Aim: To find solution of ordinary differential equation(BVP) by Finite Difference Method

Formula Used: $$\begin{pmatrix}
-2+\beta_1 d^2 & 1+\frac{\alpha_1 d}{2} & 0 & 0 &0 ....\\
1-\frac{\alpha_2 d}{2} & -2+\beta_2 d^2 & 1+\frac{\alpha_2 d}{2}& 0 &0 ...\\
0& 1-\frac{\alpha_3 d}{2} & -2+\beta_3 d^2 &1+\frac{\alpha_3d}{2} & 0..\\
. &.&.&.&.\\
.&.&.&.&.\\
0&.&.&1-\frac{\alpha_N d}{2}& -2+\beta_N d^2\end{pmatrix}$$ $$\begin{pmatrix}
y_0 \\y_1\\y_2\\.\\.\\y_n
\end{pmatrix}
=
\begin{pmatrix}
\gamma_1 d^2 \\\gamma_2 d^2\\\gamma_3 d^2\\.\\.\\\gamma_n d^2
\end{pmatrix}
$$

```python
#Finite Difference Method
import numpy as np

x_1=0
x_n=4
y_0=1
y_n=2
n=int(input('Enter the value of n: '))
d=(x_n-x_1)/n
alpha=2
beta=0
gamma=1
m=n-1
matrix_1=np.zeros((m,m))

for i in range(m):
    for j in range(m):
        if i==j:
            a=-2+beta*d**2
            matrix_1[i,j]=a
        elif j==i+1:
            b=1+alpha*d/2
            matrix_1[i,j]=b
        elif j==i-1:
            c=1-alpha*d/2
            matrix_1[i,j]=c

matrix_2=np.linalg.inv(matrix_1)
t=gamma*d**2
matrix_3=np.full(m,t)
matrix_3[0] -= c*y_0
matrix_3[-1] -= b*y_n
y=np.matmul(matrix_2,matrix_3)

print(f'{y} is the solution')
```

Aim: To solve BVP using shooting method

```python
# Shooting method
x_0=0
x_n=1
y_0=1
y_n=2

n=100
h=(x_n-x_0)/n

s_1=0
s_2=2

for j in range(10):
    x=x_0
    y=y_0
    z=s_1
    for i in range(n):
        k_1=h*z
        l_1=h*(y*y)

        k_2=h*(z+l_1/2)
        l_2=h*((y+k_1/2)*(y+k_1/2))

        k_3= h*(z+l_2/2)
        l_3=h*((y+k_2/2)*(y+k_2/2))

        k_4=h*(z+l_3)
        l_4=h*((y+k_3)*(y+k_3))

        y=y+(k_1+2*k_2+2*k_3+k_4)/6
        z=z+(l_1+2*l_2+2*l_3+l_4)/6

        x=x+h

    f_1=y-y_n

    x=x_0
    y=y_0
    z=s_2

    for i in range(n):
        k_1=h*z
        l_1=h*(y*y)

        k_2=h*(z+l_1/2)
        l_2=h*((y+k_1/2)*(y+k_1/2))

        k_3=h*(z+l_2/2)
        l_3=h*((y+k_2/2)*(y+k_2/2))

        k_4=h*(z+l_3)
        l_4=h*((y+k_3)*(y+k_3))

        y=y+(k_1+2*k_2+2*k_3+k_4)/6
        z=z+(l_1+2*l_2+2*l_3+l_4)/6

        x=x+h

    f_2=y-y_n

    # Secant method
    s_3=s_2-f_2*(s_2-s_1)/(f_2-f_1)

    if f_2 < 0:
        error=-f_2
    else:
        error=f_2
    if error < 0.0001:
        break

    s_1=s_2
    s_2=s_3

print("Initial slope =", s_3)
print("Final value of y =", y)
```
